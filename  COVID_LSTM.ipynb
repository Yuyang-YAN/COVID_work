```python
from sklearn.model_selection import train_test_split
import numpy as np

X_lstm2 =np.array(all_pos_neg2['feature'].tolist())
y_lstm2 =np.array(all_pos_neg2['new_label'].tolist())
nameall = all_pos_neg2['name'].tolist()
```

```python
import pandas as pd
import numpy as np


X =  pd.DataFrame(X_lstm2)
y = pd.DataFrame(y_lstm2)
name = pd.DataFrame(nameall)
```

```python
from sklearn.preprocessing import RobustScaler, normalize


targets = y.to_numpy().reshape(-1, 1)

RS = RobustScaler()
X_set = RS.fit_transform(X)
X  = X_set.reshape(-1,  1, X_set.shape[-1])

```

```python
inputs = X
targets =targets
```

```python
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Conv2D, MaxPooling2D, Embedding
from tensorflow.keras.losses import sparse_categorical_crossentropy
from tensorflow.keras.optimizers import Adam
from sklearn.model_selection import KFold
import numpy as np
from tensorflow.keras.models import save_model

from keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from keras.layers import *
from tensorflow.keras.optimizers import SGD

from sklearn.model_selection import KFold, StratifiedKFold
from sklearn.metrics import auc
from sklearn.metrics import confusion_matrix
from sklearn.metrics import roc_auc_score
from sklearn.metrics import roc_curve
import matplotlib.pylab as plt


from scipy import interp




# Define per-fold score containers
acc_per_fold = []
auc_per_fold = []
loss_per_fold = []
modelsss = []

kfold = StratifiedKFold(n_splits=10, shuffle=True,random_state=0)

# K-fold Cross Validation model evaluation
fold_no = 1

modelss = {}
tprs = []
aucs = []
mean_fpr = np.linspace(0,1,100)

for train, test in kfold.split(inputs, targets):
    # print("TRAINS",inputs[train],targets[train])
    # print("TEST",inputs[test],targets[test])
  # Define the model architecture
    print ('Creating model...')
    model = Sequential()

    model.add(Bidirectional(LSTM(128, return_sequences=True)))
    model.add(Dropout(0.3))
    model.add((Dense(128, activation='selu')))
    model.add(Dropout(0.3))
    model.add((Dense(64, activation='selu')))
    model.add(Dropout(0.3))
    model.add(Dense(1,activation='sigmoid'))
    #model.add(Dense(1))

    #print(model.summary())
    print ('Compiling...')
    model.compile(loss='mae',
                  optimizer='adam',
                  metrics=['accuracy'])
      # Generate a print
    print('------------------------------------------------------------------------')
    print(f'Training for fold {fold_no} ...')

    num_epochs = 100
    num_batch_size = 32
    # Fit data to model
    history = model.fit(inputs[train], targets[train],
              batch_size=num_batch_size,
              epochs=num_epochs,
              verbose=1)
    # Generate generalization metrics
    scores = model.evaluate(inputs[test], targets[test], verbose=0)

    #print("\n", classification_report(y_pred, y_test))
    pred = model.predict(inputs[test])
    y_pred = [np.round(x) for x in pred]
    test_preds1 = [np.round(x) for x in y_pred]
    test_preds1 = np.array(test_preds1)
    test_preds1 = test_preds1.flatten('A')

    testtargets1 = np.array(targets[test])
    testtargets1 = testtargets1.flatten('A')

    print('confusion matrix')
    print(confusion_matrix(testtargets1, test_preds1))

    test_preds12 = np.array(pred)
    test_preds12 = test_preds12.flatten('A')


    fpr2, tpr2, thresholds2 = roc_curve(testtargets1, test_preds12)
    AUC_lstm21 = auc(fpr2, tpr2)
    print(f'auc: {AUC_lstm21}')

    tprs.append(interp(mean_fpr, fpr2, tpr2))
    roc_auc = auc(fpr2, tpr2)
    aucs.append(roc_auc)

    plt.plot(fpr2, tpr2, lw=2, alpha=0.3, label='ROC fold %d (AUC = %0.2f)' % (fold_no, AUC_lstm21))


    print(f'Score for fold {fold_no}: {model.metrics_names[0]} of {scores[0]}; {model.metrics_names[1]} of {scores[1]*100}%')
    acc_per_fold.append(scores[1] * 100)
    auc_per_fold.append(AUC_lstm21)
    loss_per_fold.append(scores[0])
    modelsss.append(model)
    for index,acc in enumerate(acc_per_fold):
        modelss[modelsss[index]] = acc
    print(modelss)

    # Increase fold number
    fold_no = fold_no + 1
# == Provide average scores ==
print('------------------------------------------------------------------------')
print('Score per fold')
for i in range(0, len(acc_per_fold)):
    print('------------------------------------------------------------------------')
    print(f'> Fold {i+1} - Loss: {loss_per_fold[i]} - AUC: {auc_per_fold[i]} - Accuracy: {acc_per_fold[i]}%')
print('------------------------------------------------------------------------')
print('Average scores for all folds:')
print(f'> Accuracy: {np.mean(acc_per_fold)} (+- {np.std(acc_per_fold)})')
print(f'> Loss: {np.mean(loss_per_fold)}')
print(f'> AUC: {np.mean(auc_per_fold)} (+- {np.std(auc_per_fold)})')
print('------------------------------------------------------------------------')

plt.plot([0,1],[0,1],linestyle = '--',lw = 2,color = 'black')
mean_tpr = np.mean(tprs, axis=0)
mean_auc = auc(mean_fpr, mean_tpr)
plt.plot(mean_fpr, mean_tpr, color='blue',
         label=r'Mean ROC (AUC = %0.2f )' % (mean_auc),lw=2, alpha=1)
```
